# **Problem 1 TF-IDF**

## 📝**Problem Description**

Implement TF-IDF using using Python, Numpy, Pandas and whatever text cleaning library required.

The tf–idf is the product of two statistics, term frequency and inverse document frequency. There are various ways for determining the exact values of both statistics, you can use the following formulas.

### Term Frequency
$$tf_{t,d} = \log_{10}(count(t,d) +1)$$

* $tf_{t,d}$ is the frequency of the word t in the
document d

### Inverse Document Frequency
$$idf_t = \log_{10}(\frac{N}{df_t})$$

* $N$ is the total number of documents
* $df_t $ is the number of documents in which term t occurs

### TF-IDF
$$tf\text{-}idf_{t,d} = tf_{t,d} \times idf_t $$

Our goal is to get the TF-IDF vector for a given word.

## ✅**Requirements**

1. Function Definitions.
- Function 1 - `compute_tfidf_weights(train_docs)`
 - Input: A list of documents (each document is a list of words).
 - Output:
   - A TF matrix (how frequent each word appears in each document): each row corresponds to a document, each column to a word's TF value.
   - An IDF vector (how rare each word is across the documents): each entry corresponds to a word's IDF value.
   
- Function 2 - `word_tfidf_vector(word, tf_df, idf_df)`
 - Input: A word, the TF matrix (`tf_df`), and the IDF vector (`idf_df`).
 - Output: A TF-IDF vector for the input word across all documents.

2. Computation of TF matrix.
3. Computation of IDF vector.
4. Computation of the word TF-IDF vector for "cat".

````python
import numpy as np
import pandas as pd
from collections import Counter
import math

# ===========================
# === [Requirement 1: Function Implementations] ===
# ===========================

# Function 1: Compute TF and IDF matrices.
def compute_tfidf_weights(train_docs):
    """
    train_docs: List of documents (each document is a list of words)
    Returns:
        tf_df: DataFrame containing Term Frequencies (TF)
        idf_df: Series containing Inverse Document Frequencies (IDF)
    """
    # Build vocabulary
    vocab = sorted(set(term for doc in train_docs for term in doc))
    N = len(train_docs)

    # Compute TF
    tf_matrix = []
    for doc in train_docs:
        word_counts = Counter(doc)
        tf_row = []
        for word in vocab:
            count = word_counts[word]
            tf = math.log10(count + 1) if count > 0 else 0
            tf_row.append(tf)
        tf_matrix.append(tf_row)

    tf_df = pd.DataFrame(tf_matrix, columns=vocab)

    # Compute IDF
    df_counts = {word: sum(word in doc for doc in train_docs) for word in vocab}
    idf_values = [math.log10(N / df_counts[word]) if df_counts[word] > 0 else 0 for word in vocab]
    idf_df = pd.Series(idf_values, index=vocab, name="IDF")

    return tf_df, idf_df

# Function 2: Compute word-level TF-IDF vector.
def word_tfidf_vector(word, tf_df, idf_df):
    """
    word: word to compute TF-IDF for
    tf_df: Term Frequency DataFrame
    idf_df: Inverse Document Frequency Series
    Returns:
        tfidf_vector: TF-IDF vector (1 value per document)
    """
    if word not in tf_df.columns or word not in idf_df:
        # Return zero vector if word is not in vocabulary
        return pd.Series([0.0] * len(tf_df), name=word)

    tf_col = tf_df[word]
    idf_value = idf_df[word]
    tfidf_vector = tf_col * idf_value

    return tfidf_vector
````

````python
# ==============================
# Import Necessary Libraries
# ==============================
import numpy as np
import pandas as pd
import re
import torch
import torch.nn as nn
import gensim.downloader as api
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score
````

````python
# ==============================
# Load Pre-trained Word Embeddings (GloVe 300D)
# ==============================

# Apply GloVe 300 because the higher value the better accuracy of model
word_vectors = api.load("glove-wiki-gigaword-300")
embedding_dim = 300
````

````python
# ===========================================
# [Requirement 1: Preprocessing Tweets]
# ===========================================

def preprocess_tweet(tweet):
    tweet = tweet.lower()
    tweet = re.sub(r'https?://\S+|www\.\S+', '', tweet)
    tweet = re.sub(r'@\w+', '', tweet)
    tweet = re.sub(r'#', '', tweet)
    tweet = re.sub(r'[^\w\s]', '', tweet)
    return tweet.strip()
````

````python
# ===========================================
# [Requirement 2: Generate Tweet Embeddings]
# ===========================================

def tweet_to_vec(tweet, embeddings, dim=300):
    words = preprocess_tweet(tweet).split()
    vectors = [embeddings[word] for word in words if word in embeddings]
    return np.mean(vectors, axis=0) if vectors else np.zeros(dim)

X_train_emb = np.vstack([tweet_to_vec(tweet, word_vectors) for tweet in df_train['tweet']])
X_test_emb = np.vstack([tweet_to_vec(tweet, word_vectors) for tweet in df_test['tweet']])
y_train_emb = df_train['label'].values
y_test_emb = df_test['label'].values
````

## LSTM

````python
tweet_lengths = df_train['tweet'].apply(lambda x: len(preprocess_tweet(x).split()))
max_sequence_length = int(tweet_lengths.quantile(0.90))  # 90th percentile
print(f"Using max sequence length: {max_sequence_length}")
````

````python
from tensorflow.keras.utils import Sequence

class KerasTweetGenerator(Sequence):
    def __init__(self, texts, labels, embeddings, max_seq_len=40, embedding_dim=300, batch_size=32, preprocess_fn=None):
        self.texts = texts
        self.labels = labels
        self.embeddings = embeddings
        self.max_seq_len = max_seq_len
        self.embedding_dim = embedding_dim
        self.batch_size = batch_size
        self.preprocess_fn = preprocess_fn

    def __len__(self):
        return int(np.ceil(len(self.texts) / self.batch_size))

    def __getitem__(self, idx):
        batch_texts = self.texts[idx * self.batch_size: (idx + 1) * self.batch_size]
        batch_labels = self.labels[idx * self.batch_size: (idx + 1) * self.batch_size]

        # Initialize X_batch with the correct shape
        X_batch = np.zeros((len(batch_texts), self.max_seq_len, self.embedding_dim), dtype=np.float32)

        for i, tweet in enumerate(batch_texts):
            words = self.preprocess_fn(tweet).split() if self.preprocess_fn else tweet.split()
            # Get valid word vectors
            vectors = [self.embeddings[word] for word in words if word in self.embeddings]

            # Fallback: if no valid embeddings found, use a single zero vector
            if not vectors:
                vectors = [np.zeros(self.embedding_dim)]

            # Convert the list of vectors to a NumPy array
            vectors_array = np.array(vectors)

            # Apply padding or truncation to the vectors_array
            if vectors_array.shape[0] < self.max_seq_len:
                # Pad with zero vectors
                padding = np.zeros((self.max_seq_len - vectors_array.shape[0], self.embedding_dim), dtype=np.float32)
                padded_vectors = np.vstack([vectors_array, padding])
            else:
                # Truncate
                padded_vectors = vectors_array[:self.max_seq_len, :]

            # Assign the padded/truncated vectors to the X_batch slice
            X_batch[i, :, :] = padded_vectors

        return X_batch, np.array(batch_labels)
````

````python
# Set reasonable values
max_seq_len = 40
# Change embedding_dim to match the loaded GloVe vectors (300)
embedding_dim = 300
batch_size = 64

from sklearn.model_selection import train_test_split

# Split df_train into training and validation sets
train_texts, val_texts, train_labels, val_labels = train_test_split(
    df_train['tweet'], df_train['label'],
    test_size=0.1, random_state=42
)

# Create dataframes for train and val
df_train_split = pd.DataFrame({'tweet': train_texts, 'label': train_labels})
df_val = pd.DataFrame({'tweet': val_texts, 'label': val_labels})

# Initialize generators with the correct embedding_dim (300)
train_gen = KerasTweetGenerator(
    df_train_split['tweet'].tolist(), df_train_split['label'].values,
    word_vectors, max_seq_len=max_seq_len,
    embedding_dim=embedding_dim, # Use the corrected embedding_dim
    batch_size=batch_size,
    preprocess_fn=preprocess_tweet
)

val_gen = KerasTweetGenerator(
    df_val['tweet'].tolist(), df_val['label'].values,
    word_vectors, max_seq_len=max_seq_len,
    embedding_dim=embedding_dim, # Use the corrected embedding_dim
    batch_size=batch_size,
    preprocess_fn=preprocess_tweet
)

# Also update the test_gen initialization if it was intended
test_gen = KerasTweetGenerator(
    df_test['tweet'].tolist(), df_test['label'].values,
    word_vectors, max_seq_len=max_seq_len,
    embedding_dim=embedding_dim, # Use the corrected embedding_dim
    batch_size=batch_size,
    preprocess_fn=preprocess_tweet
)
````

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# Ensure lstm_units and output_dim are defined earlier
lstm_units = 64      # Example value, adjust as needed
output_dim = 5       # Number of classes

model_keras = Sequential([
    # Use the corrected embedding_dim (300) in the input_shape
    LSTM(lstm_units, input_shape=(max_sequence_length, embedding_dim)),
    Dropout(0.5),
    Dense(lstm_units // 2, activation='relu'),
    Dropout(0.5),
    Dense(output_dim, activation='softmax')
])

model_keras.compile(optimizer='adam',
                    loss='sparse_categorical_crossentropy',
                    metrics=['accuracy'])

model_keras.summary()
````

````python
epochs = 10

history = model_keras.fit(
    train_gen,
    validation_data=val_gen,
    epochs=epochs,
    verbose=1
)
````

````python
# Predict on test set
y_pred_probs_keras = model_keras.predict(test_gen)
y_pred_keras = np.argmax(y_pred_probs_keras, axis=1)
y_test_keras = df_test['label'].values

# Evaluation
from sklearn.metrics import accuracy_score, f1_score

print("\n===== Classifier 2 (Neural Network - Keras LSTM) =====")
print(f"Accuracy (Micro): {accuracy_score(y_test_keras, y_pred_keras):.4f}")
print(f"Accuracy (Macro): {accuracy_score(y_test_keras, y_pred_keras):.4f}")
print(f"F1 Score (Micro): {f1_score(y_test_keras, y_pred_keras, average='micro'):.4f}")
print(f"F1 Score (Macro): {f1_score(y_test_keras, y_pred_keras, average='macro'):.4f}")
````

````python
# Helper functions
def tokenize_and_align_labels(examples):
    tokenized_inputs = tokenizer(examples["tokens"], truncation=True, is_split_into_words=True)

    labels = []
    for i, label in enumerate(examples[f"upos"]):
        word_ids = tokenized_inputs.word_ids(batch_index=i)
        previous_word_idx = None
        label_ids = []
        for word_idx in word_ids:
            # Special tokens have a word id that is None. We set the label to -100 so they are automatically
            # ignored in the loss function.
            if word_idx is None:
                label_ids.append(-100)
            # We set the label for the first token of each word.
            elif word_idx != previous_word_idx:
                label_ids.append(label[word_idx])
            # For the other tokens in a word, we set the label to either the current label or -100, depending on
            # the label_all_tokens flag.
            else:
                label_ids.append(label[word_idx] if label_all_tokens else -100)
            previous_word_idx = word_idx

        labels.append(label_ids)

    tokenized_inputs["labels"] = labels
    return tokenized_inputs

from sklearn.metrics import f1_score, accuracy_score, precision_score, recall_score
def compute_metrics(pred):
    labels = pred.label_ids
    preds = pred.predictions.argmax(-1)

    # Masking ignored indices (e.g., -100) used during training
    true_labels = [l for p, l in zip(preds.flatten(), labels.flatten()) if l != -100]
    pred_labels = [p for p, l in zip(preds.flatten(), labels.flatten()) if l != -100]

    return {
        "accuracy": accuracy_score(true_labels, pred_labels),
        "f1": f1_score(true_labels, pred_labels, average='macro'),
        "precision": precision_score(true_labels, pred_labels, average='macro'),
        "recall": recall_score(true_labels, pred_labels, average='macro'),
    }
````

````python
import numpy as np
from datasets import load_dataset, Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForTokenClassification,
    DataCollatorForTokenClassification,
    TrainingArguments,
    Trainer
)

import os

# --- !!! ADD THIS LINE AT THE TOP !!! ---
os.environ["WANDB_DISABLED"] = "true"
# ---

# ======================
# === [Requirement 1: Data Loading & Preprocessing Function Implementation] ===
# ======================

# Configuration
MODEL_CHECKPOINT = "distilbert-base-cased"
DATASET_NAME = "universal_dependencies"
DATASET_SUBSET = "en_ewt"

# Load the dataset
raw_datasets = load_dataset(DATASET_NAME, DATASET_SUBSET)
print(raw_datasets)

# Get label list (Universal POS tags)
label_list = raw_datasets["train"].features[f"upos"].feature.names
print(f"\nFound {len(label_list)} POS tags: {label_list}")

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)
model = AutoModelForTokenClassification.from_pretrained(
    MODEL_CHECKPOINT,
    num_labels=len(label_list),
    id2label={i: label for i, label in enumerate(label_list)},
    label2id={label: i for i, label in enumerate(label_list)}
)
````

````python
# ======================
# === [Requirement 2: Label Alignment & Tokenization Logic] ===
# ======================

# Set label_all_tokens flag (False means only label first subword token)
label_all_tokens = False

# Apply tokenization and label alignment to all datasets
tokenized_datasets = raw_datasets.map(tokenize_and_align_labels, batched=True)
````

````python
# ======================
# === [Requirement 3: Training Setup & Execution] ===
# === You need to test on the test set, hint: compute_metrics(trainer.predict(tokenized_datasets["test"])) ===
# ======================

# Set up data collator for token classification
data_collator = DataCollatorForTokenClassification(tokenizer)

# Define training arguments
training_args = TrainingArguments(
    output_dir="./results",
    eval_strategy="epoch",
    save_strategy="epoch",
    logging_dir="./logs",
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    learning_rate=5e-5,
    logging_steps=100,
    load_best_model_at_end=True,
    metric_for_best_model="eval_accuracy",
    greater_is_better=True,
    report_to=None,  # Disable external tracking
    save_total_limit=2,
)

# Initialize trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["validation"],
    tokenizer=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
)

# Start training
print("\nStarting training...")
trainer.train()

# Evaluate on test set
print("\nEvaluating on test set...")
test_results = trainer.predict(tokenized_datasets["test"])
test_metrics = compute_metrics(test_results)

print("\nTest dataset metrics:")
print(test_metrics)
````
