#### Import modules and Dataset

```python
# Import modules

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from keras.models import Sequential
from keras.layers import Dense
import numpy as np
```

```python
# Import dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Preprocess data
# Flatten the images
x_train = x_train.reshape((x_train.shape[0], 28 * 28))  
x_test = x_test.reshape((x_test.shape[0], 28 * 28))      

# Normalize
x_train = x_train.astype('float32') / 255               
x_test = x_test.astype('float32') / 255

# Encode the outputs with one-hot encoding
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
```

## Task 1

### Model 1 – Basic Neural Network

```python
# Create NN model
model_1= models.Sequential()
model_1.add(layers.Dense(512, activation='relu', input_shape=(28 * 28,)))  
model_1.add(layers.Dense(256, activation='relu'))                          
model_1.add(layers.Dense(10, activation='softmax'))                       

# Compile the model
model_1.compile(loss='categorical_crossentropy', optimizer='sgd', metrics=['accuracy'])

# Train the model
tf.random.set_seed(1)
np.random.seed(1)
model_1.fit(x_train, y_train, epochs=10, batch_size=128, validation_split=0.5) # Use 50% of training dataset for validation

# Evaluate the model
test_loss, test_acc = model_1.evaluate(x_test, y_test)
print(f'Test accuracy model 1: {test_acc}')
```

### Model 1a – NN - improved by adding 2 more dense layers and changing the number of neurons in dense layers

```python
# Create NN model
model_1a = models.Sequential()
model_1a.add(layers.Dense(512, activation='relu', input_shape=(28 * 28,)))  
model_1a.add(layers.Dense(256, activation='relu'))                          
model_1a.add(layers.Dense(128, activation='relu'))
model_1a.add(layers.Dense(64, activation='relu'))  
model_1a.add(layers.Dense(10, activation='softmax'))                        

# Compile the model
model_1a.compile(loss='categorical_crossentropy', optimizer='sgd', metrics=['accuracy'])

# Train the model
tf.random.set_seed(1)
np.random.seed(1)
model_1a.fit(x_train, y_train, epochs=10, batch_size=128, validation_split=0.5) # Use 50% of training dataset for validation

# Evaluate the model
test_loss, test_acc = model_1a.evaluate(x_test, y_test)
print(f'Test accuracy model 1a: {test_acc}')
```

### Model 2 – NN  – improved by adding one more dense layer and changing optimizer

```python
# Create NN model
model_2 = models.Sequential()
model_2.add(layers.Dense(512, activation='relu', input_shape=(28 * 28,)))  
model_2.add(layers.Dense(256, activation='relu'))                          
model_2.add(layers.Dense(128, activation='relu'))                          
model_2.add(layers.Dense(10, activation='softmax'))                        

# Compile the model
model_2.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# Train the model
tf.random.set_seed(1)
np.random.seed(1)
model_2.fit(x_train, y_train, epochs=10, batch_size=128, validation_split=0.5) # Use 50% of training dataset for validation

# Evaluate the model
test_loss, test_acc = model_2.evaluate(x_test, y_test)
print(f'Test accuracy model 2: {test_acc}')
```

## Task 2

```python
# Import modules

from keras.layers import Dropout
from keras.constraints import maxnorm
from tensorflow.keras.optimizers import SGD
from keras.layers.convolutional import Conv2D
from keras.layers.convolutional import MaxPooling2D
from keras.layers import Flatten
```

```python
tf.random.set_seed(1) 
np.random.seed(1)

# Conv2D layers
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# Encode the outputs with one-hot encoding
num_classes = y_test.shape[1]
```

### Model 3 - CNN – with activation relu and softmax in the convolutional layers

```python
# Build the model 3
model_3 = Sequential()
model_3.add(Conv2D(32, (3, 3), input_shape=(28, 28, 1), padding='same', activation='relu', kernel_constraint=maxnorm(3)))
model_3.add(Conv2D(32, (3, 3), activation='softmax', padding='same', kernel_constraint=maxnorm(3)))
model_3.add(Conv2D(64, (3, 3), activation='relu', padding='same', kernel_constraint=maxnorm(3)))
model_3.add(MaxPooling2D())
model_3.add(Flatten())
model_3.add(Dense(512, activation='relu', kernel_constraint=maxnorm(3)))
model_3.add(Dropout(0.2))
model_3.add(Dense(num_classes, activation='softmax'))
```

```python
# Define optimizer
learning_rate = 0.002 
epochs = 5
decay = learning_rate / epochs
sgd = SGD(lr=learning_rate, momentum=0.7, decay=decay, nesterov=False)

# Compile the model
model_3.compile(loss='categorical_crossentropy', optimizer=sgd, metrics=['accuracy'])
```

```python
# Train the model
epochs = 5
batch_size = 60

# Fit the model
model_3.fit(x_train, y_train, validation_split=0.5, epochs=epochs, batch_size=batch_size, verbose=1)
```

```python
# Evaluate the trained model

scores = model_3.evaluate(x_test, y_test, verbose=0)
print("Accuracy model 3: %.2f%%" % (scores[1]*100))
```

### Model 4 – CNN – adding one more convolutional layer and setting relu activation function for all convolutional layers

```python
# Build the model 4
model_4 = Sequential()
model_4.add(Conv2D(32, (3, 3), input_shape=(28, 28, 1), padding='same', activation='relu', kernel_constraint=maxnorm(3)))
model_4.add(Conv2D(32, (3, 3), activation='relu', padding='same', kernel_constraint=maxnorm(3)))
model_4.add(Conv2D(64, (3, 3), activation='relu', padding='same', kernel_constraint=maxnorm(3)))
model_4.add(MaxPooling2D())
model_4.add(Flatten())
model_4.add(Dense(512, activation='relu', kernel_constraint=maxnorm(3)))
model_4.add(Dropout(0.2))
model_4.add(Dense(num_classes, activation='softmax'))
```

```python
# Define optimizer
learning_rate = 0.002 
epochs = 5
decay = learning_rate / epochs
sgd = SGD(lr=learning_rate, momentum=0.7, decay=decay, nesterov=False)

# Compile the model
model_4.compile(loss='categorical_crossentropy', optimizer=sgd, metrics=['accuracy'])
```

```python
# Train the model
epochs = 5
batch_size = 60

# Fit the model
model_4.fit(x_train, y_train, validation_split=0.5, epochs=epochs, batch_size=batch_size, verbose=1)
```

```python
# Evaluate the trained model

scores = model_4.evaluate(x_test, y_test, verbose=0)
print("Accuracy model 4: %.2f%%" % (scores[1]*100))
```
